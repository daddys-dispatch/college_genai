```python
import gensim.downloader as ptwv
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
```

```python
print("Loading pre-trained embeddings...")

model = ptwv.load("glove-wiki-gigaword-100")
# model = ptwv.load("word2vec-google-news-300")

print("Model loaded successfully")
```

```python
words = [
    "computer",
    "internet",
    "software",
    "hardware",
    "keyboard",
    "mouse",
    "server",
    "network",
    "programming",
    "database",
]

vectors = [model[word] for word in words]
```

```python
pca = PCA(n_components=2)
reduced = pca.fit_transform(vectors)
```

```python
input_word = "computer"

similar_words = model.most_similar(
    positive=[input_word],
    topn=5
)

print(f"Top 5 words similar to '{input_word}':\n")

for word, score in similar_words:
    print(f"{word}: {score:.4f}")
```

```python
plt.figure(figsize=(8, 6))

for i, word in enumerate(words):
    plt.scatter(reduced[i, 0], reduced[i, 1])
    plt.annotate(word, (reduced[i, 0], reduced[i, 1]))

plt.title("PCA Visualization of Technology Word Embeddings")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)

plt.show()
```

## Analysis

1. Words such as **server**, **network**, and **database** tend to appear close together because they are related to networking and backend systems.

2. **Keyboard** and **mouse** may form a separate cluster since they are hardware input devices.

3. **Computer**, **software**, and **programming** are often positioned near each other because they frequently occur in similar technical contexts.

4. PCA reduces hundreds of dimensions into two dimensions, so some semantic information is lost. However, related words still tend to remain close together.

5. The generated similar words demonstrate that embeddings capture semantic meaning by placing related concepts near one another in vector space.
