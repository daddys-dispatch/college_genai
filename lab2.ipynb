Analyze clusters and relationships. Generate contextually rich outputs using embeddings. Generate 5 semantically similar words for a given input.

```python
# Import required modules
import gensim.downloader as ptwv
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
```

```python
# Load pre-trained model
print("Loading Word2Vec model...")
model = ptwv.load("word2vec-google-news-300")
print("Model loaded successfully")
```

```python
# Select 10 domain-specific words (Technology)
words = [
    "computer",
    "internet",
    "software",
    "hardware",
    "keyboard",
    "mouse",
    "server",
    "network",
    "programming",
    "database",
]
vectors = [model[word] for word in words]
```

```python
# Dimensionality reduction using PCA
pca = PCA(n_components=2)
reduced = pca.fit_transform(vectors)
```

```python
# Generate 5 semantically similar words for given input
input_word = "computer"
similar_words = model.most_similar(positive=[input_word], topn=5)

print(f"Top 5 words similar to '{input_word}':\n")
for word, score in similar_words:
    print(f"{word}: {score:.4f}")
```

```python
# Plot the word embeddings
plt.figure(figsize=(8, 6))
for i, word in enumerate(words):
    plt.scatter(reduced[i, 0], reduced[i, 1])
    plt.annotate(word, (reduced[i, 0], reduced[i, 1]))
plt.title("PCA Visualization of Technology Word Embeddings")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True)
plt.show()
```
